```python
import pandas as pd
```

```python
data = {
    'Produk': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headset',
               'Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headset'],
    'Kategori': ['Elektronik', 'Aksesoris', 'Aksesoris', 'Elektronik', 'Aksesoris',
                 'Elektronik', 'Aksesoris', 'Aksesoris', 'Elektronik', 'Aksesoris'],
    'Harga': [7000000, 150000, 300000, 2500000, 500000,
              7000000, 150000, 300000, 2500000, 500000],
    'Jumlah': [2, 5, 3, 2, 4, 1, 8, 5, 3, 6]
}

df = pd.DataFrame(data)

df
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

df['Produk'].value_counts().plot(kind='bar')

plt.title('Jumlah Penjualan Berdasarkan Produk')
plt.xlabel('Produk')
plt.ylabel('Jumlah Terjual')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(df['Produk'], df['Harga'])

plt.title('Harga Produk')
plt.xlabel('Produk')
plt.ylabel('Harga (Rp)')

plt.show()
```

```python
produk_terlaris = df.groupby('Produk')['Jumlah'].sum()

produk_terlaris
```

```python
produk_terlaris.idxmax()
```

```python
df['Total_Pendapatan'] = df['Harga'] * df['Jumlah']

df
```

```python
total_pendapatan = df['Total_Pendapatan'].sum()

print("Total Pendapatan: Rp", total_pendapatan)
```

```python
pendapatan_produk = df.groupby('Produk')['Total_Pendapatan'].sum()

pendapatan_produk.plot(kind='bar', figsize=(8,5))

plt.title('Total Pendapatan Berdasarkan Produk')
plt.xlabel('Produk')
plt.ylabel('Total Pendapatan (Rp)')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby('Kategori')['Total_Pendapatan'].sum()
```

```python
df.groupby('Kategori')['Total_Pendapatan'].sum().plot(kind='bar')

plt.title('Pendapatan Berdasarkan Kategori')
plt.xlabel('Kategori')
plt.ylabel('Total Pendapatan')
plt.show()
```

```python
df.sort_values('Total_Pendapatan', ascending=False).head()
```

```python
pendapatan_produk.sort_values().plot(kind='barh', figsize=(8,5))

plt.title('Total Pendapatan per Produk')
plt.xlabel('Pendapatan')
plt.ylabel('Produk')
plt.show()
```
